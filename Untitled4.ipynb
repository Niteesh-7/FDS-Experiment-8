```python
import pandas as pd

regions = pd.read_csv("retail_regions.txt")
customers = pd.read_csv("retail_customers.txt")
products = pd.read_csv("retail_products.txt")
```

```python
print("REGIONS")
print(regions)

print("\nCUSTOMERS")
print(customers)

print("\nPRODUCTS")
print(products)
```

```python
print("Regions Shape:", regions.shape)
print("Customers Shape:", customers.shape)
print("Products Shape:", products.shape)
```

```python
filtered_data = customers[
    customers["Region"] == "West"
]

print(filtered_data)

filtered_sales = regions[
    regions["Total_Sales"] > 100000
]

print(filtered_sales)

filtered_data = regions[
    (regions["Total_Sales"] > 100000) &
    (regions["Total_Profit"] > 10000)
]

print(filtered_data)
```

```python
sorted_data = regions.sort_values(
    by="Total_Sales",
    ascending=False
)

print(sorted_data)
```

```python
part1 = customers.iloc[:5]
part2 = customers.iloc[5:10]
combined_data = pd.concat(
    [part1, part2],
    ignore_index=True
)

print(combined_data)
```

```python
merged_data = pd.merge(
    customers,
    regions,
    on="Region",
    how="inner"
)

print(merged_data)
```

```python
merged_all = pd.merge(
    merged_data,
    products,
    how="cross"
)

print(merged_all)
```

```python
summary = merged_all.groupby(
    "Region"
)["Total_Sales"].agg(
    ["sum", "mean", "min", "max"]
)

print(summary)
```

```python
reshaped_data = pd.pivot_table(
    merged_all,
    values="Total_Sales",
    index="Region",
    columns="Category",
    aggfunc="sum"
)

print(reshaped_data)
```

```python
reshape_data = merged_all[
    [
        "Region",
        "Customer_ID",
        "Customer_Name",
        "Product_ID",
        "Product_Name",
        "Category",
        "Sub_Category",
        "Total_Sales",
        "Total_Profit",
        "Total_Quantity"
    ]
]
reshaped_data = reshape_data.set_index(
    ["Region", "Category"]
)
reshaped_data = reshaped_data.reset_index()

print(reshaped_data)
```
